## Алхам 2: Гэр бүлийн мэдлэгийн сан (баримт)

```python
families = [
    (('bataa', 'saraa'), ['bold', 'oyuna', 'tomor']),
    (('bold', 'tsetseg'), ['anar', 'zolboo']),
    (('oyuna', 'dorj'), ['munkh', 'naraa']),
    (('tomor', 'ganaa'), ['enkhee']),
    (('anar', 'bat'), ['temuulen']),
]
for couple, kids in families:
    for p in couple:
        for c in kids:
            +parent(p, c)
for name in ['bataa', 'bold', 'tomor', 'dorj', 'zolboo', 'munkh', 'temuulen', 'bat']:
    +male(name)
for name in ['saraa', 'oyuna', 'tsetseg', 'ganaa', 'anar', 'naraa', 'enkhee']:
    +female(name)
```

```python
def count(query):
    result = pyDatalog.ask(query)
    return 0 if result is None else len(result.answers)

print(count("parent(X, Y)"), count("male(X)"), count("female(X)"))
print(pyDatalog.ask("parent('bataa', X)").answers)
```

## Алхам 3: Энгийн дүрэм

```python
father(X, Y) <= parent(X, Y) & male(X)
mother(X, Y) <= parent(X, Y) & female(X)
son(X, Y) <= parent(Y, X) & male(X)
daughter(X, Y) <= parent(Y, X) & female(X)
person(X) <= male(X)
person(X) <= female(X)
```

```python
print(pyDatalog.ask("father(X, 'anar')").answers)
print(pyDatalog.ask("mother(X, 'bold')").answers)
print(pyDatalog.ask("son(X, 'oyuna')").answers)
print(pyDatalog.ask("daughter(X, 'oyuna')").answers)
print(count("father(X, Y)"), count("mother(X, Y)"), count("person(X)"))
```

## Алхам 4: Өвөө эмээ ба ах дүү

```python
grandparent(X, Y) <= parent(X, Z) & parent(Z, Y)
sibling(X, Y) <= parent(P, X) & parent(P, Y) & (X != Y)
sibling_any(X, Y) <= parent(P, X) & parent(P, Y)
```

```python
print(pyDatalog.ask("grandparent(X, 'temuulen')").answers)
print(count("grandparent(X, Y)"))
print(pyDatalog.ask("sibling(X, 'munkh')").answers)
print(pyDatalog.ask("sibling(X, 'bold')").answers)
print(count("sibling(X, Y)"))
print(count("sibling_any(X, Y)"))
```

## Алхам 5: Ураг төрлийн дүрмүүд

```python
brother(X, Y) <= sibling(X, Y) & male(X)
sister(X, Y) <= sibling(X, Y) & female(X)
uncle(X, Y) <= parent(P, Y) & brother(X, P)
aunt(X, Y) <= parent(P, Y) & sister(X, P)
cousin(X, Y) <= parent(P, X) & parent(Q, Y) & sibling(P, Q)
```

```python
print(pyDatalog.ask("brother(X, 'oyuna')").answers)
print(pyDatalog.ask("sister(X, 'zolboo')").answers)
print(pyDatalog.ask("uncle(X, 'anar')").answers)
print(pyDatalog.ask("aunt(X, 'anar')").answers)
print(pyDatalog.ask("uncle(X, 'temuulen')").answers)
print(pyDatalog.ask("cousin(X, 'enkhee')").answers)
print(pyDatalog.ask("cousin(X, 'anar')").answers)
print(count("cousin(X, Y)"))
```

## Алхам 6: Рекурсив дүрэм

```python
ancestor(X, Y) <= parent(X, Y)
ancestor(X, Y) <= parent(X, Z) & ancestor(Z, Y)
descendant(X, Y) <= ancestor(Y, X)
```

```python
print(pyDatalog.ask("ancestor('bataa', 'temuulen')") is not None)
print(pyDatalog.ask("ancestor(X, 'temuulen')").answers)
print(pyDatalog.ask("descendant(X, 'bataa')").answers)
print(pyDatalog.ask("ancestor(X, 'anar')").answers)
print(count("ancestor(X, Y)"))
```

## Алхам 7: Асуулт ба үгүйсгэл

```python
has_child(X) <= parent(X, Y)
childless(X) <= person(X) & ~has_child(X)
great_grandparent(X, Y) <= parent(X, Z) & grandparent(Z, Y)
```

```python
print(count("childless(X)"))
print(pyDatalog.ask("childless('bataa')") is not None, pyDatalog.ask("childless('temuulen')") is not None)
print(pyDatalog.ask("great_grandparent(X, 'temuulen')").answers)
```

### Асуултын хүснэгт

```python
print("1.", pyDatalog.ask("parent('bold', X)").answers)
print("2.", pyDatalog.ask("grandparent(X, 'temuulen')").answers)
print("3.", pyDatalog.ask("uncle(X, 'munkh')").answers)
print("4.", pyDatalog.ask("cousin(X, 'enkhee')").answers)
print("5.", sorted({a[0] for a in pyDatalog.ask("cousin(X, Y)").answers}))
print("6.", sorted({a[0] for a in pyDatalog.ask("childless(X)").answers}))
print("7.", pyDatalog.ask("great_grandparent(X, 'temuulen')").answers)
print("8.", sorted({a[0] for a in pyDatalog.ask("grandparent(X, Y)").answers}))
```

## Алхам 4 ба Алхам 8-ын үгээр хариулах асуултууд

Эцэг эхгүй bataa, saraa, bat нар нэмэгддэггүй.
